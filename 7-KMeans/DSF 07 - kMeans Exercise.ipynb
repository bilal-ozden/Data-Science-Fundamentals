## 7.1 Introduction & Motivation

Unlike regression and classification, clustering is a bit more abstract. We no longer know which value we want to predict or what kind of values to expect. Instead, we give our data to the model and ask a different question: within this data, are there any **groups** (or **clusters**) of data points that belong together?

At first glance, this kind of information might seem less valuable than a regression or classification model, but it is extremely useful in practice. Think of customer segmentation (knowing which groups of customers you have, so you know who to target), fraud detection (spotting transactions that don't really fit in any cluster) or organizing data (such as grouping similar articles).

**Key insight**: there is often no single "right" answer in clustering. The key is to make reasonable choices based on your data and goals, and to justify them.
